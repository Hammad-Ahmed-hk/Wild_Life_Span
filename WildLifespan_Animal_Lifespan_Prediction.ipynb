### 📥 Task 1 — Load the Dataset

We load the dataset directly from **Google Sheets** by exporting it as a CSV file.  
The dataset contains **200 animal species** with features like Speed, Weight, and Lifespan.

```python
# Task 1: Load dataset directly from Google Sheets (exported as CSV)
url = "https://docs.google.com/spreadsheets/d/1N4yfjXZvtQtkLfpUez78uZOZ0RD6hh8chCdkrmEglOA/export?format=csv"

df = pd.read_csv(url)

print("✅ Dataset loaded successfully!")
print(f"📐 Shape: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"\n📋 Column Names: {list(df.columns)}")
print("\n🔍 First 5 rows:")
df.head()
```

### 🧹 Task 2 — Clean the Data

We keep only the **three numerical columns** we need for our models:
- 🏃 **Speed** (km/h) — how fast the animal moves
- ⚖️ **Weight** (kg) — how heavy the animal is
- ⏳ **Lifespan** (years) — how long the animal lives *(this is what we want to predict)*

We also **drop any rows with missing values** to ensure clean, reliable data.

```python
# Task 2: Select only the 3 numerical columns we need
df_clean = df[['Speed', 'Weight', 'Lifespan']].copy()

# Drop rows that have any missing (NaN) values
df_clean.dropna(inplace=True)

print("✅ Data cleaned successfully!")
print(f"\n📊 Rows remaining after cleaning : {len(df_clean)}")
print(f"❌ Missing values per column:\n{df_clean.isnull().sum().to_string()}")
print("\n📈 Statistical Summary:")
df_clean.describe().round(2)
```

### 🌡️ Task 3 — Correlation Heatmap

A **Correlation Heatmap** shows us how strongly two variables are related to each other.

| Correlation Value | Meaning |
|:---:|:---|
| **1.0** | Perfect positive relationship |
| **0.5 to 0.9** | Strong positive relationship |
| **0.0 to 0.3** | Weak or no relationship |
| **-1.0** | Perfect negative relationship |

> 🔍 **What to look for:** Check the numbers in the **Lifespan** row/column — are Speed and Weight strongly correlated with Lifespan?

```python
# Task 3: Correlation Heatmap
plt.figure(figsize=(7, 5))
correlation = df_clean.corr()

mask = np.zeros_like(correlation, dtype=bool)

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="RdYlGn",
    linewidths=1.5,
    linecolor='white',
    square=True,
    annot_kws={"size": 14, "weight": "bold"},
    vmin=-1, vmax=1
)

plt.title("🌡️ Correlation Heatmap\nSpeed, Weight & Lifespan",
          fontsize=14, fontweight='bold', pad=15)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11, rotation=0)
plt.tight_layout()
plt.show()

# Interpretation
speed_corr  = correlation.loc['Speed',  'Lifespan']
weight_corr = correlation.loc['Weight', 'Lifespan']

print(f"\n📊 Correlation with Lifespan:")
print(f"   🏃 Speed  → Lifespan : {speed_corr:.3f}")
print(f"   ⚖️  Weight → Lifespan : {weight_corr:.3f}")

print("\n💬 Interpretation:")
print("   Values close to 0 indicate WEAK relationships.")
print("   This suggests Speed & Weight may NOT be strong Lifespan predictors.")
```

### 📊 Task 4 — Lifespan Distribution Chart

A **Distribution Chart (Histogram)** tells us:
- What is the most **common lifespan range** in our dataset?
- Is the data **normally distributed** or **skewed**?
- Are there **outliers** (extremely short or long-lived animals)?

> 🧠 This helps us understand the **target variable** we're trying to predict.

```python
# Task 4: Lifespan Distribution Chart
fig, ax = plt.subplots(figsize=(10, 5))

n, bins, patches = ax.hist(
    df_clean['Lifespan'], bins=20,
    color='#4e9af1', edgecolor='white',
    alpha=0.85, linewidth=1.2
)

# Add a KDE (smooth curve) on top
df_clean['Lifespan'].plot(kind='kde', ax=ax, secondary_y=False,
                          color='#e94560', linewidth=2.5, label='Density Curve')

mean_val   = df_clean['Lifespan'].mean()
median_val = df_clean['Lifespan'].median()

ax.axvline(mean_val,   color='red',    linestyle='--', linewidth=2, label=f'Mean   : {mean_val:.1f} yrs')
ax.axvline(median_val, color='orange', linestyle=':',  linewidth=2, label=f'Median : {median_val:.1f} yrs')

ax.set_title("📊 Distribution of Animal Lifespan", fontsize=15, fontweight='bold', pad=12)
ax.set_xlabel("Lifespan (Years)", fontsize=12)
ax.set_ylabel("Number of Animals", fontsize=12)
ax.legend(fontsize=11)
plt.tight_layout()
plt.show()

print(f"\n📌 Key Statistics:")
print(f"   Average Lifespan : {mean_val:.2f} years")
print(f"   Median Lifespan  : {median_val:.2f} years")
print(f"   Shortest Animal  : {df_clean['Lifespan'].min()} years")
print(f"   Longest Animal   : {df_clean['Lifespan'].max()} years")
```

---
## 📏 Phase 2: Building the Linear Regression Model

**Linear Regression** assumes the relationship between features and target is a **straight line**.

The formula it learns looks like this:

$$\text{Lifespan} = \text{Intercept} + (\beta_1 \times \text{Speed}) + (\beta_2 \times \text{Weight})$$

- **Intercept** → The baseline lifespan when Speed = 0 and Weight = 0
- **β₁ (Speed Coefficient)** → For every +1 km/h of speed, lifespan changes by this value
- **β₂ (Weight Coefficient)** → For every +1 kg of weight, lifespan changes by this value

---

### ✂️ Task 5 — Split the Data (80% Train / 20% Test)

We split the dataset into two parts:

| Split | Purpose | Size |
|:---|:---|:---:|
| **Training Set** | The model *learns* from this data | 80% |
| **Testing Set** | We test how well the model *generalizes* to unseen data | 20% |

> 🎯 This is like studying (training) before an exam (testing) — the model should never see the test data before evaluation!

```python
# Task 5: Define Features (X) and Target (y)
X = df_clean[['Speed', 'Weight']]   # 🔢 Input features
y = df_clean['Lifespan']            # 🎯 Target to predict

# Split: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,    # 20% for testing
    random_state=42    # Seed for reproducibility
)

print("✅ Data split successfully!")
print(f"\n📦 Total samples   : {len(df_clean)}")
print(f"   🎓 Training set  : {len(X_train)} samples ({len(X_train)/len(df_clean)*100:.0f}%)")
print(f"   🧪 Testing set   : {len(X_test)} samples  ({len(X_test)/len(df_clean)*100:.0f}%)")
```

### 🏋️ Task 6 & 7 — Train Linear Model & Inspect Coefficients

Now we train the **Linear Regression** model and look at what it learned:

> 🔑 The **Coefficients** reveal how the model *thinks* Speed and Weight affect Lifespan.

```python
# Task 6: Train the Linear Regression model
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

print("✅ Linear Regression model trained successfully!")

# Task 7: Print the model parameters
print("\n" + "="*52)
print("       📐 LINEAR MODEL PARAMETERS")
print("="*52)
print(f"  📍 Intercept         : {linear_model.intercept_:.4f}")
print(f"  🏃 Speed Coefficient : {linear_model.coef_[0]:.4f}")
print(f"  ⚖️  Weight Coefficient: {linear_model.coef_[1]:.4f}")
print("="*52)

print("\n💡 Formula Learned:")
print(f"   Lifespan = {linear_model.intercept_:.2f}")
print(f"             + ({linear_model.coef_[0]:.4f} × Speed)")
print(f"             + ({linear_model.coef_[1]:.4f} × Weight)")
print("\n📌 A positive coefficient means the feature INCREASES lifespan")
print("   A negative coefficient means the feature DECREASES lifespan")
```

---
## 📐 Phase 3: Polynomial Regression Model

Sometimes real-world relationships **aren't straight lines** — they curve!

**Polynomial Regression (Degree 2)** adds **squared versions** of the features:
$$\text{Lifespan} = \beta_0 + \beta_1 \cdot \text{Speed} + \beta_2 \cdot \text{Weight} + \beta_3 \cdot \text{Speed}^2 + \beta_4 \cdot \text{Speed} \cdot \text{Weight} + \beta_5 \cdot \text{Weight}^2$$

This gives the model **more flexibility** to fit curved patterns in the data.

---

### 🔄 Task 8 & 9 — Transform Features & Train Polynomial Model

**Step 1:** Use `PolynomialFeatures` to create new derived features (Speed², Weight², Speed×Weight, etc.)  
**Step 2:** Train a `LinearRegression` model on these **transformed features**

> 💡 We're still using *Linear Regression* under the hood — we just feed it **richer features**!

```python
# Task 8: Apply Polynomial Feature Transformation (Degree 2)
poly_converter = PolynomialFeatures(degree=2, include_bias=False)

X_train_poly = poly_converter.fit_transform(X_train)  # Fit on train, transform train
X_test_poly  = poly_converter.transform(X_test)       # Only transform test (no fitting!)

feature_names = poly_converter.get_feature_names_out(['Speed', 'Weight'])

print("✅ Polynomial transformation done!")
print(f"\n   Original features   : {X_train.shape[1]}")
print(f"   Transformed features: {X_train_poly.shape[1]}")
print(f"\n   New feature set: {list(feature_names)}")
print("   (Includes: Speed, Weight, Speed², Speed×Weight, Weight²)")

# Task 9: Train Polynomial Regression model
poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)

print("\n✅ Polynomial Regression model trained successfully!")
```

---
## 📊 Phase 4: Model Accuracy Report

Now comes the most important question: **Which model performed better?**

We use the **R² Score (R-Squared)** to measure accuracy:

| R² Score | Interpretation |
|:---:|:---|
| **1.00** | 🏆 Perfect model — predicts everything correctly |
| **0.70+** | ✅ Good model |
| **0.30–0.70** | ⚠️ Moderate model |
| **0.00–0.30** | ❌ Weak model — features are poor predictors |
| **< 0.00** | 🚨 Worse than just guessing the average |

---

### 🏆 Task 10 — Calculate R² Scores

```python
# Task 10: Generate predictions from both models
y_pred_linear = linear_model.predict(X_test)
y_pred_poly   = poly_model.predict(X_test_poly)

# Calculate R² Scores
r2_linear = r2_score(y_test, y_pred_linear)
r2_poly   = r2_score(y_test, y_pred_poly)

print("\n" + "="*52)
print("          📊 MODEL ACCURACY REPORT")
print("="*52)
print(f"  📏 Linear Regression     R² = {r2_linear:.4f}")
print(f"  📐 Polynomial Regression R² = {r2_poly:.4f}")
print("="*52)

# Verdict
if r2_poly > r2_linear:
    diff = r2_poly - r2_linear
    print(f"\n🏆 Polynomial Model wins by {diff:.4f} R² points")
elif r2_linear > r2_poly:
    diff = r2_linear - r2_poly
    print(f"\n🏆 Linear Model wins by {diff:.4f} R² points")
else:
    print("\n🤝 Both models performed equally")

print("\n💡 Note: Scores near 0 suggest Speed & Weight are poor predictors of Lifespan.")
```

### 🎯 Task 11 — Actual vs Predicted Scatter Plot

The **Actual vs Predicted** scatter plot is one of the most useful diagnostic tools in ML:

- **Each dot** represents one animal from the test set
- **X-axis** = the actual real lifespan
- **Y-axis** = what the model predicted
- **Red dashed line** = the "Perfect Guess" line — if a dot lies on it, the prediction was exactly right!
- **Dots far from the line** = the model made a large error there

> 🎯 A great model has most dots **clustered tightly around the red dashed line**.

```python
# Task 11: Actual vs Predicted Scatter Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

def plot_actual_vs_predicted(ax, y_actual, y_predicted, title, r2, color):
    ax.scatter(y_actual, y_predicted, color=color, alpha=0.65,
               edgecolors='white', linewidths=0.6, s=70, label='Predictions')

    # Perfect guess line
    lims = [min(y_actual.min(), y_predicted.min()) - 2,
            max(y_actual.max(), y_predicted.max()) + 2]
    ax.plot(lims, lims, 'r--', linewidth=2.2, label='Perfect Guess Line', zorder=5)

    ax.set_xlim(lims)
    ax.set_ylim(lims)
    ax.set_title(f"{title}\nR² = {r2:.4f}", fontsize=13, fontweight='bold', pad=10)
    ax.set_xlabel("Actual Lifespan (years)", fontsize=11)
    ax.set_ylabel("Predicted Lifespan (years)", fontsize=11)
    ax.legend(fontsize=10)
    ax.set_aspect('equal')

plot_actual_vs_predicted(axes[0], y_test, y_pred_linear,
                         "📏 Linear Regression", r2_linear, '#4e9af1')

plot_actual_vs_predicted(axes[1], y_test, y_pred_poly,
                         "📐 Polynomial Regression", r2_poly, '#f4845f')

plt.suptitle("🎯 Actual vs Predicted Lifespan — Model Comparison",
             fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("\n📌 How to read this plot:")
print("   • Dots ON the red line  = perfect prediction")
print("   • Dots FAR from the line = large prediction error")
print("   • Scattered dots        = model is struggling to find a pattern")
```

## 💾 Phase 5: Saving Models for Deployment

Once a model is trained, we **save it to disk** so it can be used later without retraining.  
We use Python's built-in `pickle` library for this.

| File | Contents |
|:---|:---|
| `linear_model.pkl` | The trained Linear Regression model |
| `poly_model.pkl` | The trained Polynomial Regression model |
| `poly_converter.pkl` | The PolynomialFeatures transformer (needed to process new input) |

> ⚠️ **Important:** To make a prediction with the polynomial model later, you **must** also load `poly_converter.pkl` and transform your input data first!

---

### 💾 Task 12 — Save Models with Pickle

```python
# Task 12: Save all 3 model files using pickle

# 1. Save Linear Regression Model
with open('linear_model.pkl', 'wb') as f:
    pickle.dump(linear_model, f)
print("✅ Saved: linear_model.pkl")

# 2. Save Polynomial Regression Model
with open('poly_model.pkl', 'wb') as f:
    pickle.dump(poly_model, f)
print("✅ Saved: poly_model.pkl")

# 3. Save the Polynomial Feature Converter
with open('poly_converter.pkl', 'wb') as f:
    pickle.dump(poly_converter, f)
print("✅ Saved: poly_converter.pkl")

print("\n" + "="*42)
print("  📁 All 3 model files saved!")
print("="*42)
print("  📦 linear_model.pkl")
print("  📦 poly_model.pkl")
print("  📦 poly_converter.pkl")
print("="*42)
print("\n🚀 Models are ready for deployment!")
```

### ✅ Verify Models — Load & Test a Prediction

Let's confirm the saved models load correctly by making a **sample prediction**:

> 🐘 **Test Animal:** Speed = 40 km/h, Weight = 5000 kg  
> What lifespan will our models predict?

```python
# Verify: Load all 3 saved model files
with open('linear_model.pkl', 'rb') as f:
    loaded_linear = pickle.load(f)

with open('poly_model.pkl', 'rb') as f:
    loaded_poly = pickle.load(f)

with open('poly_converter.pkl', 'rb') as f:
    loaded_converter = pickle.load(f)

print("✅ All 3 models loaded successfully from disk!")

# --- Test Prediction ---
# Sample: Speed=40 km/h, Weight=5000 kg (think of an elephant!)
sample = pd.DataFrame([[40, 5000]], columns=['Speed', 'Weight'])

linear_pred = loaded_linear.predict(sample)[0]
poly_pred   = loaded_poly.predict(loaded_converter.transform(sample))[0]

print("\n" + "="*48)
print("   🐾 SAMPLE PREDICTION")
print("   Animal: Speed=40 km/h | Weight=5,000 kg")
print("="*48)
print(f"   📏 Linear Model Prediction    : {linear_pred:.2f} years")
print(f"   📐 Polynomial Model Prediction: {poly_pred:.2f} years")
print("="*48)
```
